```python
import os
import time
import pandas as pd
from woc.remote import WocMapsRemote

woc = WocMapsRemote(base_url='https://worldofcode.org/api/')
assigned = pd.read_csv('net2prj.csv')
projects = list(assigned[assigned['netID'] == 'jtyuill']['WoC'])

if os.path.exists('jtyuill_project_summary.csv'):
    # already downloaded, reuse it instead of pulling every commit again
    summary = pd.read_csv('jtyuill_project_summary.csv', sep=';')
else:
    rows = []
    for project in projects:
        shas = woc.get_values('p2c', project)
        for start in range(0, len(shas), 10):
            results, errors = woc.get_values_many('commit.tch', shas[start:start + 10])
            for sha, value in results.items():
                tree, parents, author, committer, message = value[0]
                rows.append({
                    'project_wocid': project,
                    'commit_sha1': sha,
                    'author': author[0],
                    'time': int(author[1]),
                    'commit message': message,
                })
            time.sleep(1)
    summary = pd.DataFrame(rows)
    summary.to_csv('jtyuill_project_summary.csv', sep=';', index=False)

summary['project_wocid'].value_counts()
```

```python
stats = []
for project in projects:
    commits = summary[summary['project_wocid'] == project]
    stats.append({
        'project': project,
        'commits': len(commits),
        'authors': commits['author'].nunique(),
        'min time': commits['time'].min(),
        'max time': commits['time'].max(),
    })
pd.DataFrame(stats)
```
